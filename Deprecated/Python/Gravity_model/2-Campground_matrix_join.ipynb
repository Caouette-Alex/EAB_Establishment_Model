```python
#import CSV into a dataframe for distance and cleaned campground data
Route_distance_df=pd.read_csv('Route_distance_calc.csv')
Campground_df=pd.read_csv('Campground_Origin_Cleand.csv',index_col=0)
Campground_df.head(1)
```

```python
Route_distance_df
```

```python
#combine data from campground data with distance in km from origin to campground 
Camp_dist_df=pd.merge(left=Campground_df, right=Route_distance_df, left_on='CODE',right_on='Name')
```

```python
#remove unnecessary columns and rename
Camp_dist_df=Camp_dist_df.drop(columns=['OID','ObjectID','Name','FirstStopI','LastStopID','StopCount'],axis=1)
Camp_dist_df=Camp_dist_df.rename(columns={'Total_Leng':'Distance_Km'})
Camp_dist_df.head(1)
```

```python
#write to csv file in OneDrive - University of New Brunswick/Grad Studies/Modelling_transport_pathways/Clean_data_sets
Camp_dist_df.to_csv('Campground_Dist_Clean.csv')
```

```python
#Insert a new column with population data from each origin location
Camp_dist_df.insert(14,'Population',0)
Population=[]
for index,row in Camp_dist_df.iterrows():
    if row['Origin'] == 'Edmunston':
        population='12086'
    elif row['Origin'] == 'Bedford':
        population='403131'
    elif row['Origin'] == 'Oromocto':
        population='9223'
    elif row['Origin'] == 'Moncton':
        population='71889'
    elif row['Origin'] == 'Fredericton':
        population='58220'
    Population.append(population)
Camp_dist_df['Population']=Population
```

```python
#Remove any values that don't include real numbers to simplify(this can be removed when the campground data sheet is cleaned up)
import numpy as np
Camp_dist_df=Camp_dist_df.replace('?',np.NaN)
Camp_dist_df=Camp_dist_df.replace(r'^\s*$',np.NaN,regex=True)
Camp_dist_df=Camp_dist_df.replace('>50',np.NaN)
Camp_dist_df=Camp_dist_df.replace('>45',np.NaN)
Camp_dist_df=Camp_dist_df.replace('>60',np.NaN)
```

```python
#Fill NAN sites to have zero 
Camp_dist_df['Sites'] = Camp_dist_df['Sites'].fillna(0)
Camp_dist_df
```

```python
Camp_dist_df=Camp_dist_df.astype({'Sites':'float64'})
```

```python
Camp_dist_df.dtypes
```

```python
Camp_dist_piv_df=Camp_dist_df.pivot(index='Campground', columns='Origin', values=['Distance_Km','Sites','Population'])
```
