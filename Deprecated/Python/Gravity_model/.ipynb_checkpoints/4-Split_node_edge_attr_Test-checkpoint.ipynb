```python
import pandas as pd
import os 

os.chdir('/..')
os.chdir('Users/alexp/OneDrive - University of New Brunswick/Grad Studies/Modelling_transport_pathways/Clean_data_sets')
os.getcwd()
```

```python
df=pd.read_csv('campground_origin_data_raw.csv',index_col=0)
df_2=pd.read_csv('campground_cleand.csv')

import numpy as np
#drop NA or 0 values
df= df.replace('?',np.NaN).replace(r'^\s*$',np.NaN,regex=True).replace('>50',np.NaN).replace('>45',np.NaN).replace('>60',np.NaN).dropna(subset=['Sites'])
df= df[df['Sites'] != 0]

#Drop any nan values
df_2= df_2.replace('?',np.NaN).replace(r'^\s*$',np.NaN,regex=True).replace('>50',np.NaN).replace('>45',np.NaN).replace('>60',np.NaN).dropna(subset=['Sites'])


edge_df= df.drop(columns=['Sites','Population'])
node_df= df.drop(columns=['Distance_Km'])
edge_df.to_csv('edge_dataframe.csv')
```

```python
#Get df of population
pop_node_df=node_df[:5]
pnode_df=pop_node_df.drop(columns=['Campground','Sites']).rename(columns={'Origin':'Name'})
```

```python
#Get df of campgrounds
cnode_df=df_2[['Name','Sites']]    
```

```python
#merge data together and change data types to include only 0 values
pnode_df=pnode_df.append(cnode_df).reset_index().drop(columns=['index'])
pnode_df ['Population']= pnode_df['Population'].fillna(0).astype(int)
pnode_df['Sites'] = pnode_df['Sites'].fillna(0).astype(int)
pnode_df.to_csv('node_dataframe.csv')
```
