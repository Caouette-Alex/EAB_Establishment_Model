```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
```

```python
Gravity_df=pd.read_csv('Gravity_norm.csv',index_col=0)
```

```python
Gravity_df=Gravity_df.astype(float)
```

```python
np.log(Gravity_df).plot.hist(grid=True, bins=20, rwidth=0.9)
plt.title('Log transformed introduction values for a matrix')
plt.xlabel('log relative probability')
plt.ylabel('Counts')
plt.grid(axis='y', alpha=0.75)
```
