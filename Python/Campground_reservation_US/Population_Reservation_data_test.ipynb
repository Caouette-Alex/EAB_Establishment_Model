## Calculating the number of campsite reservations made in a city
### To do this I needed to:
#### 1. Clean the dataset to only include cities in the state of Maine
#### 2. Calculate the number of campground reservations made at each exclusive zip code
#### 3. Join all the zip codes to fit into a single city and sum the number of reservations

```python
import pandas as pd
```

```python
#importing the csv containing historical reservation data from the Unites States for the 2020 camping season(FY 2020), Website URL: https://ridb.recreation.gov/download
Reservation_df=pd.read_csv('FY20_Historical_Reservations_Full.csv')
```

```python
#print data
Reservation_df
```

```python
##Discontinued coding line went with people from origin location not destination location
#including only **camping** reservations from the state of Maine 
#Maine_df = Reservation_df.loc[Reservation_df['facilitystate'] == 'Maine'].loc[Reservation_df['inventorytype'] == 'CAMPING']
#Maine_df.to_csv('Maine_reservations.csv')
```

```python
#Include only camping reservations from the origin State of Maine
Maine_origin_df = Reservation_df.loc[Reservation_df['customerstate'] == 'ME'].loc[Reservation_df['inventorytype'] == 'CAMPING']
Maine_origin_df.to_csv('origin_state.csv')
```

```python
#Take the zip codes and convert them into a single list
test_df = Maine_origin_df['customerzip'].value_counts().index.tolist()
```

```python
#add a '0' in front of each zip code so that it matches the format from zipcodes
Test_list=[]
for x in test_df:
    zip_code='0'+ x
    Test_list.append(zip_code)
```

```python
#Summarize the number of campground reservations there were for a specific ZIP code.
sum_df = Maine_origin_df['customerzip'].value_counts().rename_axis('zip').reset_index(name='counts')
```

```python
#add a '0' in front of each zip code so that it matches the format from zipcodes and change the dataframe column data to match the change
sum_list=[]
for x in sum_df['zip']:
    bingo= '0'+ x
    sum_list.append(bingo)
sum_df['zip']= sum_list
```

```python
#add data from the reservation zip codes to add extra data like city, state, etc... 
Maine_test_zip=[]
import zipcodes
for x in Test_list:
    Maine_test_zip.append(zipcodes.matching(x))
```

```python
#remove unnecessary square brackets outside of the dictionary: from: [[{dict1}],[{dict2}],[{dict3}]] to:[{dict1},{dict2},{dict3}]
flattened = [] 
for sublist in Maine_test_zip: 
    for val in sublist: 
        flattened.append(val) 
```

```python
#convert to a dataframe
df= pd.DataFrame(flattened,columns=('zip_code','zip_code_type','active','city','acceptable_cities','unacceptable_cities','state','county','timezone','area_codes','world_region','country','lat','long'))
```

```python
#drop non-important columns
df.drop(columns=['timezone','world_region','country'])
```

```python
#Merge zip code data(df) to the campground reservation counts(sum_df) 
merge_df=pd.merge(left=df, right=sum_df, left_on='zip_code', right_on='zip')
```

```python
#remove any zip codes for businesses/large groups and for PO Boxes
merge_df = merge_df[merge_df.zip_code_type == 'STANDARD']
```

```python
#Rename count column to better represent actual value: number of reservations
merge_df = merge_df.rename(columns={'counts':'#_of_reservations'})
```

```python
merge_df
```

```python
#Combine zip codes within the same city limits and adding the number of reservations together
test_reserv_count = merge_df.groupby(['city','state'], as_index=False)['#_of_reservations'].sum()
```

```python
#add new column with Maine as the state to measure
test_reserv_count['State'] = 'Maine'
```

```python
# test to see if the number  of reservations summed equals that of Portland when measured individually
merge_df.loc[merge_df['city'] == 'Portland']
```

```python
#test to see if the number of reservations matches the sum for a city like portland
test_reserv_count.sort_values(by='#_of_reservations')
```

```python
test_reserv_count.to_csv('Maine_reservations_by_city.csv')
```
