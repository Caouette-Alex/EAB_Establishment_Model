```python
Zip_code_list= Reservation_df['customerzip'].value_counts().index.tolist()
```

```python
Zip_code=[]
for code in Zip_code_list:
    if len(code) == 4:
        zip_code='0'+ code
        Zip_code.append(zip_code)
    else:
        Zip_code.append(code)
```

```python
for code in Zip_code:
    if len(code) < 5:
        print('Bad code')
    elif len (code) > 6:
        print('Bad Code')
    
```

```python
Reservation_sum_df = Reservation_df['customerzip'].value_counts().rename_axis('zip').reset_index(name='counts')
```

```python
Reservation_sum_list=[]
for zcode in Reservation_sum_df['zip']:
    if len(zcode) == 4:
        zip_code='0'+ zcode
        Reservation_sum_list.append(zip_code)
    else:
        Reservation_sum_list.append(zcode)
Reservation_sum_df['zip'] = Reservation_sum_list
```

```python
for zcode in Reservation_sum_df['zip']:
    if len(zcode) < 5:
        print('Zip code to short')
        print(zcode)
    elif len (zcode) > 6:
        print('Zip code too long')
```

```python
Zip_code_info=[]
import zipcodes
for x in Zip_code:
    Zip_code_info.append(zipcodes.matching(x))
```

```python
flattened = [] 
for sublist in Zip_code_info: 
    for val in sublist: 
        flattened.append(val)
```

```python
#convert to a dataframe
Zip_code_df = pd.DataFrame(flattened,columns=('zip_code','zip_code_type','active','city','acceptable_cities','unacceptable_cities','state','county','timezone','area_codes','world_region','country','lat','long'))
```
