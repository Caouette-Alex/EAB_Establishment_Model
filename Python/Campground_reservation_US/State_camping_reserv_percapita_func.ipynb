'''Get mean and SD reservations data from national park reservations systems in the US

```python
import pandas as pd
import os
import numpy as np

os.chdir ('/..')
os.chdir('Users/alexp/OneDrive - University of New Brunswick/Grad Studies/Modelling_transport_pathways/Clean_data_sets')
os.getcwd()
```

```python

Pop_df=pd.read_csv('Merged_pop_reserv_all_more_5.csv',index_col=0)

#Create a nested dictionary of dataframes for each state in the US
State_df={}
states = sorted(Pop_df['STNAME'].unique())
for state in states:
    State_df[state+'_df'] = Pop_df[Pop_df.STNAME == state]
```

```python
#Calculate the mean and standard deviation of each state dataframe's per capita rate of reservations to campgrounds
State_res_variation={}
Mean_percap_df=pd.DataFrame()
for k, v in State_df.items():
    mean=np.mean(v['Per_capita_rate'])
    median=np.median(v['Per_capita_rate'])
    SD=np.std(v['Per_capita_rate'])
    State_res_variation[k] = mean, median, SD
    #print(k, '=',  mean, u"\u00B1", SD)  
Mean_percap_df = Mean_percap_df.from_dict(State_res_variation, orient='index',columns=['Mean','Median', 'Standard_dev'])
    
```

```python
list_mean=[]
for percap in Mean_percap_df['Mean']:
    list_mean.append(percap)
list_SD=[] 
for percap in Mean_percap_df['Standard_dev']:
    list_SD.append(percap)
list_median=[] 
for percap in Mean_percap_df['Median']:
    list_median.append(percap)
```
