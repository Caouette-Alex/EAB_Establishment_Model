```python
import pandas as pd
```

```python
US_df = pd.read_csv('US_city_population.csv')
```

```python
US_df
```

```python
US_df = US_df[['NAME','STNAME','POPESTIMATE2019']]
```

```python
NAME_short= US_df['NAME'].str.rstrip('city').str.rstrip('town').str.rstrip('County')
```

```python
NAME_short
```

```python
NAME_list=[]
for name in NAME_short:
    x=name
    NAME_list.append(x)  
```

```python
US_df.replace(to_replace='NAME', value=NAME_list,regex=True)
```
