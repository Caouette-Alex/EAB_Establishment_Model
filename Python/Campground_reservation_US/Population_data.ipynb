## Clean data up by removing unnecessary columns and excess words in the city andtown names

```python
import pandas as pd
```

```python
# Import US city population data from the US on the Census website using the data from:https://www.census.gov/data/tables/time-series/demo/popest/2010s-total-cities-and-towns.html#ds
US_df = pd.read_csv('US_city_population.csv')
```

```python
US_df
```

```python
#keep only the city/town name, state name, and the population estimate in 2019
US_df = US_df[['NAME','STNAME','POPESTIMATE2019']]
```

```python
#Remove the word city and town from the end of the place name and remove the spaces at the end of the doucuments
NAME_short= US_df['NAME'].str.rstrip('city').str.rstrip('town').str.rstrip()
```

```python
#convert it to a list because otherwise it's in some weird format
NAME_list=[]
for name in NAME_short:
    x=name
    NAME_list.append(x)
```

```python
#readd the city name into the dataframe
US_df['Name'] = NAME_list
```

```python
US_df.drop(columns=['NAME'])
```

```python
US_df.loc[US_df['Name'] == 'Abbot']
```

```python
US_df= US_df[['Name','STNAME','POPESTIMATE2019']]
```

```python
US_df
```

```python
# Drop place names that have duplicated names
US_df = US_df.drop_duplicates(subset=['Name','STNAME'])
```

```python
US_df.to_csv('US_city_pop_clean.csv')
```
