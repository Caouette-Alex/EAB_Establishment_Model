## Looking for the lowest probability of introduction and highest probability of introduction sites using the sum of gravity values from all origin locations

```python
#import packages and read csv of gravity matrix and read csv of root(pop) gravity matrix

import pandas as pd

Gravity = pd.read_csv('Gravity_calc.csv', index_col=0)
Norm_gravity = pd.read_csv('Gravity_norm.csv',index_col=0)
```

```python
#Summing the probability values for all 5 origin locations
Gravity_sum = Gravity['Bedford']+ Gravity['Edmunston']+ Gravity['Fredericton']+ Gravity['Moncton']+ Gravity['Oromocto']

Norm_gravity_sum = Norm_gravity['Bedford']+ Norm_gravity['Edmunston']+ Norm_gravity['Fredericton']+ Norm_gravity['Moncton']+ Norm_gravity['Oromocto']
```

```python
#Sort values from smallest to largest
Gravity_sum = Gravity_sum.sort_values()
Norm_gravity_sum = Norm_gravity_sum.sort_values()
```
