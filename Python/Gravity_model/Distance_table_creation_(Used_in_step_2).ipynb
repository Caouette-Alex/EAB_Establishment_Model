# Merge **two** datasheets
---
### This code merges the dataframes of two datasets using a column that is similar between the two  
1. To do this we import pandas  
2. Open both .csv files as dataframes
3. merge dataframes
4. remove unnecessary columns

```python
#Importing packages for analysis
import pandas as pd
import os 

os.chdir("/..")
os.chdir("Users/alexp/OneDrive - University of New Brunswick/Grad Studies/Modelling_transport_pathways/Test_data")
print("Directory changed")
```

```python
#Import distance table and route table to
routedist_df=pd.read_csv('Route_distance_output.csv')
routename_df=pd.read_csv('Route_table.csv')
```

```python
#merge two datasets using campground code number and origin code number
distmerge_df=pd.merge(left=routename_df, right=routedist_df, left_on='Code_Col', right_on='Name')
```

```python
#remove unnecessary columns and rename lat, long, and code columns
distclean_df=distmerge_df.drop(['OID', 'C_Latitude', 'O_Latitude','FirstStopID', 'LastStopID','StopCount','Name','OBJECTID','Column1'], axis=1).rename(columns={'C_Longitud':'C_long', 'O_Longitud':'O_long','Code':'Camp_Code','Column2':'Origin_Code'})
#reorder columns
distclean_df=distclean_df[['Campground','Origin','Province','C_Lat','C_long','O_Lat','O_long','Camp_Code','Origin_Code','Code_Col','Total_Length','Shape_Length']]
distclean_df
```

```python
#Shift the columns to show distance of each origin location to each campground in meters
distpivot_df = distclean_df.pivot(index='Campground', columns='Origin', values='Total_Length')
distpivot_df
```
